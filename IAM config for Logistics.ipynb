```python
%sql
-- Grant USE CATALOG so the group can see the catalog exists
GRANT USE CATALOG ON CATALOG logistics_lakehouse TO `innowise`;

-- Grant USE SCHEMA on the schemas
GRANT USE SCHEMA ON SCHEMA logistics_lakehouse.bronze TO `innowise`;
GRANT USE SCHEMA ON SCHEMA logistics_lakehouse.silver TO `innowise`;
GRANT USE SCHEMA ON SCHEMA logistics_lakehouse.gold TO `innowise`;

-- Grant SELECT permissions on all tables within the schemas
GRANT SELECT ON SCHEMA logistics_lakehouse.bronze TO `innowise`;
GRANT SELECT ON SCHEMA logistics_lakehouse.silver TO `innowise`;
GRANT SELECT ON SCHEMA logistics_lakehouse.gold TO `innowise`;

-- Unity Catalog uses a positive grant model - the group will only have
-- the permissions explicitly granted (SELECT, USE SCHEMA, USE CATALOG).
-- No MODIFY permission means they cannot write to the data.
```

```python
%sql
-- Check all grants on catalog
SHOW GRANTS ON CATALOG logistics_lakehouse;

-- Check grants on each schema
SHOW GRANTS ON SCHEMA logistics_lakehouse.bronze;
SHOW GRANTS ON SCHEMA logistics_lakehouse.silver;
SHOW GRANTS ON SCHEMA logistics_lakehouse.gold;
```
