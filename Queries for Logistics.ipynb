```python
%sql 
-- Query the Gold Layer (DQL)

-- Compare average revenue by driver experience level
SELECT 
    d.experience_level,
    COUNT(t.trip_id) AS total_trips,
    ROUND(AVG(t.fare_amount), 2) AS avg_fare,
    ROUND(AVG(t.fare_per_mile), 2) AS avg_fare_per_mile
FROM logistics_lakehouse.gold.fact_trips t
JOIN logistics_lakehouse.gold.dim_drivers d ON t.driver_id = d.driver_id
GROUP BY d.experience_level
ORDER BY avg_fare DESC;
```

```python
%sql
-- Query the Gold Layer (DQL)

-- Filter for premium drivers driving modern vehicles
SELECT 
    d.name,
    v.model,
    v.vehicle_age_category
FROM logistics_lakehouse.gold.fact_trips t
JOIN logistics_lakehouse.gold.dim_drivers d ON t.driver_id = d.driver_id
JOIN logistics_lakehouse.gold.dim_vehicles v ON t.car_number = v.car_number
WHERE d.is_premium = True AND v.vehicle_age_category = 'Modern Fleet';
```
