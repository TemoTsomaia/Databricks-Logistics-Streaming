```python
%sql
CREATE CATALOG IF NOT EXISTS logistics_lakehouse;
USE CATALOG logistics_lakehouse;

CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

-- Volume for streaming checkpoints
CREATE VOLUME IF NOT EXISTS logistics_lakehouse.bronze.checkpoints;

-- Creation of Gold Tables (Simple Start Schema and DDL)
CREATE OR REPLACE TABLE logistics_lakehouse.gold.dim_drivers (
    driver_id LONG, 
    name STRING, 
    experience INTEGER,
    experience_level STRING, -- Gold Layer Exclusive 
    rating FLOAT, 
    is_premium BOOLEAN, -- Gold Layer Exclusive
    phone STRING,
    loaded_at TIMESTAMP -- Gold Layer Exclusive
) CLUSTER BY (driver_id);

CREATE OR REPLACE TABLE logistics_lakehouse.gold.dim_vehicles (
    car_number STRING, 
    model STRING, 
    year INTEGER,
    vehicle_age_category STRING, -- Gold Layer Exclusive
    status STRING,
    loaded_at TIMESTAMP -- Gold Layer Exclusive
);

CREATE OR REPLACE TABLE logistics_lakehouse.gold.fact_trips (
    trip_id STRING, 
    driver_id LONG, 
    car_number STRING, 
    distance_miles FLOAT, 
    fare_amount FLOAT,
    fare_per_mile FLOAT, -- Gold Layer Exclusive
    loaded_at TIMESTAMP -- Gold Layer Exclusive
) CLUSTER BY (driver_id, car_number);
```
